## Data Transformation Exercises (Encoding & Scaling)


### Setup

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

# Load dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Cleaning steps from Session 3
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)

embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

df = df.drop(columns=['Cabin'])

# Cap Fare outliers
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
df.loc[df['Fare'] > upper_bound, 'Fare'] = upper_bound

df.head()
```

## Exercise 1 — Encoding Categorical Data


### 1. One-Hot Encode Nominal Data: Sex & Embarked

```python
df_encoded = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
df_encoded.head()
```

### 2. Label Encoding Ordinal Data: Pclass

```python
le = LabelEncoder()
df_encoded['Pclass'] = le.fit_transform(df_encoded['Pclass'])
df_encoded[['Pclass']].head()
```

```python
df_encoded['Pclass'].value_counts()
```

## Exercise 2 — Scaling Numerical Data


### 1. Standardization — StandardScaler

```python
scaler = StandardScaler()
scaled_std = scaler.fit_transform(df[['Age','Fare']])

df_std = pd.DataFrame(scaled_std, columns=['Age_scaled','Fare_scaled'])
df_std.describe()
```

### 2. Normalization — MinMaxScaler

```python
minmax = MinMaxScaler()
scaled_mm = minmax.fit_transform(df[['Age','Fare']])

df_mm = pd.DataFrame(scaled_mm, columns=['Age_norm','Fare_norm'])
df_mm.describe()
```

## Exercise 3 — The Transformation Strategist (Conceptual)
### Recommended Methods:
- **neighborhood** → One-hot encoding (nominal categorical)
- **num_rooms** → No encoding needed (numerical); scaling optional
- **building_condition** → Ordinal encoding (Poor < Average < Good < Excellent)
- **price_amd** → StandardScaler or MinMaxScaler depending on model

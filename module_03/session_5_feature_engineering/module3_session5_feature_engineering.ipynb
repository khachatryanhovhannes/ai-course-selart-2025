Ready-to-use notebook.

```python

import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Cleaning from Session 3
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df.drop('Cabin', axis=1, inplace=True)

df.head()
```

## Exercise 1: Numerical Feature Engineering

```python

# FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# IsAlone
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

df[['SibSp','Parch','FamilySize','IsAlone']].head()
```

## Exercise 2: Extracting Titles

```python

df['Title'] = df['Name'].str.extract(r'([A-Za-z]+)\.')

df['Title'].value_counts()
```

```python

rare_titles = ['Dr','Rev','Col','Major','Mlle','Countess','Ms','Lady','Sir','Jonkheer','Don','Dona','Capt']
df['Title'] = df['Title'].replace(rare_titles, 'Rare')

df['Title'].value_counts()
```

## Exercise 3: Age Binning

```python

bins = [0,12,18,60,100]
labels = ['Child','Teen','Adult','Senior']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

df[['Age','AgeGroup']].head(10)
```

```python
df['AgeGroup'].value_counts()
```

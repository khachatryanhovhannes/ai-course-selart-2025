## Exercise 1: Conceptual Questions

### Foundation

This session compares sparse count-based representations such as TF-IDF with dense semantic representations such as word embeddings.

### Build

1. **TF-IDF Intuition**

I would expect **`explosion`** to have a higher average IDF score than **`Tuesday`**. `Tuesday` appears in many news articles because reporting often references dates, schedules, and timelines, so it is less distinctive across the corpus. `Explosion` is more topic-specific and appears in a narrower set of articles, which makes it more informative and pushes its IDF higher.

2. **Vector Similarity**

Cosine similarity is usually preferred for TF-IDF vectors because it compares direction rather than raw magnitude. Longer documents naturally contain more words, which can inflate Euclidean distance even when two documents discuss the same topic. Cosine similarity largely removes that document-length effect and focuses on how similar the term distributions are.

3. **Embeddings vs. TF-IDF**

**Word embeddings** would be better for recognizing the similarity between `feline` and `kitten`. TF-IDF treats them as separate surface forms unless they literally co-occur in similar documents, while embeddings are trained to place semantically related words close together in vector space. That makes embeddings much more suitable for semantic recommendation tasks where related words may differ in form.

### Result

TF-IDF is strong for measuring term importance inside a corpus, while embeddings are stronger when semantic similarity matters across different word choices.


## Exercise 2: TF-IDF with Scikit-learn

This exercise builds a TF-IDF matrix for a very small corpus and inspects the learned vocabulary.


````python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    'the quick brown fox jumps over the lazy dog',
    'a brown dog was chasing a brown cat',
    'the lazy cat sleeps all day'
]

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(corpus)
````

````python
feature_names = vectorizer.get_feature_names_out()
dense_matrix = tfidf_matrix.toarray()

print('Feature names:')
print(feature_names)
print()
print('Dense TF-IDF matrix:')
print(np.round(dense_matrix, 3))
````

### Analysis (Markdown Answer)

### Foundation

Each column in the TF-IDF matrix corresponds to one vocabulary term learned by the vectorizer.

### Build

The column for **`brown`** can be identified by checking the position of `brown` in `feature_names`. It has a non-zero value in the first and second rows because `brown` appears in the first and second documents. It is zero in the third row because the third document does not contain the word at all. Since stop words are removed, the matrix focuses on content-bearing words such as `brown`, `dog`, `cat`, and `lazy`.

### Result

TF-IDF captures both presence and importance: a word only receives weight in documents where it appears, and common words across the corpus are down-weighted relative to more distinctive ones.


## Exercise 3: Challenge Problem - Exploring Pre-trained Embeddings

### Foundation

This exercise uses a pre-trained GloVe model through `gensim` to inspect semantic neighborhoods, analogies, and similarity scores.

### Build

The first run downloads `glove-wiki-gigaword-50`, which is about 69 MB. After that, the model is loaded from the local cache.

### Result

Run the next cells to explore how the embedding space organizes meaning.


````python
import gensim.downloader as api

word_vectors = api.load('glove-wiki-gigaword-50')
````

````python
print('Most similar to woman:')
print(word_vectors.most_similar('woman', topn=5))
print()
print('Most similar to car:')
print(word_vectors.most_similar('car', topn=5))
````

````python
print("Analogy: king - man + woman")
print(word_vectors.most_similar(positive=['king', 'woman'], negative=['man'], topn=5))
print()
print("Analogy: France : Paris :: Germany : ?")
print(word_vectors.most_similar(positive=['Germany', 'Paris'], negative=['France'], topn=5))
print()
print(f"Similarity(cat, dog): {word_vectors.similarity('cat', 'dog'):.3f}")
print(f"Similarity(cat, car): {word_vectors.similarity('cat', 'car'):.3f}")
````

### Analysis (Markdown Answer)

### Foundation

Pre-trained embeddings are useful because they capture statistical patterns from very large corpora, which often translate into semantic structure.

### Build

The nearest-neighbor results for words like `woman` and `car` usually return terms that are semantically or contextually related, which is what we would expect from a meaningful embedding space. The analogy task `king - man + woman` often returns `queen` or something very close to it, and the geography analogy typically returns `Berlin` near the top. The similarity score between `cat` and `dog` is usually much higher than the score between `cat` and `car`, which matches common sense.

### Result

Yes, this demonstrates that the model has captured a useful form of meaning. It does not understand language the way a human does, but it has clearly learned patterns of relatedness, category membership, and some relational structure from large-scale text.

## Exercise 2: Few-Shot Prompting in Action (45 minutes)

### Foundation

In-context learning can teach a model output format and task behavior through examples inside the prompt, even without weight updates.

This exercise compares zero-shot, one-shot, and few-shot prompting quality for structured extraction.

### Build

Task: Convert unstructured text into structured JSON.

1. **Zero-Shot Attempt**

   Use an LLM interface and ask it to convert this text into JSON:

   "The meeting is with John Smith tomorrow at 2 PM in Conference Room 3. The topic is the Q3 budget review. Please remember to bring the latest sales figures."

   Analyze output quality:
   - Is it valid JSON?
   - Is the structure clean and consistent?

2. **One-Shot Attempt**

   Use one example in the prompt:

   ```text
   Convert the following text to JSON.
   Text: "Lunch with Jane Doe on Friday at 12:30pm at The Corner Cafe to discuss the project proposal."
   JSON: {"event": "Lunch", "person": "Jane Doe", "day": "Friday", "time": "12:30 PM", "location": "The Corner Cafe", "topic": "Project Proposal"}
   Text: "The meeting is with John Smith tomorrow at 2 PM in Conference Room 3. The topic is the Q3 budget review. Please remember to bring the latest sales figures."
   JSON:
   ```

   Compare this output with the zero-shot output.

3. **Few-Shot Attempt**

   Add one more slightly different example before the final task text, then run again.

   Analyze whether additional examples improve consistency and robustness to wording variations.

### Result

Your notebook should include prompts, outputs, and a short comparison of zero-shot vs one-shot vs few-shot behavior.

## Exercise 3: Challenge Problem - Chain-of-Thought Prompting

### Foundation

Direct-answer prompting can trigger intuitive but wrong responses on reasoning tasks.

Chain-of-Thought prompting often improves performance by encouraging intermediate reasoning steps.

### Build

Task: Solve a simple logic puzzle with and without step-by-step prompting.

1. **Standard Prompt Attempt**

   Ask:

   "A bat and a ball cost $1.10 in total. The bat costs $1.00 more than the ball. How much does the ball cost?"

   Record the answer.

   Optional extra puzzle:
   "I want to wash my car. The car wash is 50 meters away. Should I walk or drive?"

2. **Chain-of-Thought Prompt Attempt**

   Ask the same bat-and-ball question but append:

   "Let's think step by step."

   Check whether the model now shows structured reasoning and reaches the correct answer.

3. **Analysis**

   Write a short paragraph explaining why Chain-of-Thought prompting can be more effective.

   Explain how decomposing the problem into intermediate steps helps avoid intuitive but incorrect responses.

### Result

Your submission should include both prompt styles, observed outputs, and a concise reasoning-quality comparison.

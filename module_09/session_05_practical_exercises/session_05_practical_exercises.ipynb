## Exercise 1: Conceptual Questions (60 minutes)

### Foundation

Advanced GAN variants solve different generation problems: controlled generation, unpaired translation, and high-fidelity synthesis.

Understanding conditioning, consistency constraints, and dataset bias is essential for responsible GAN design.

### Build

Answer the following questions in markdown cells or comments.

1. **cGAN Design**

   You are training a Conditional GAN for clothing images. Classes are:
   - clothing type: shirt, trousers, shoes
   - color: red, green, blue

   How would you design the condition label `y` so a user can request combinations such as "blue shirt"?

   Explain how this condition should be provided to both Generator and Discriminator.

2. **CycleGAN Failure Modes**

   You are training a CycleGAN to translate summer landscape photos to winter landscapes.

   a) If you used only adversarial loss, what easy but incorrect solution might the Generator find?

   b) How does cycle consistency loss prevent this specific failure mode?

3. **StyleGAN and Bias**

   StyleGAN is trained on large face datasets (for example FFHQ).

   If the dataset is imbalanced by ethnicity, gender, or age, how would that appear in generated outputs?

   Why is this an important ethical concern in generative AI systems?

### Result

Your answers should show clear understanding of conditional control, unpaired-translation constraints, and fairness risks from dataset bias.

## Exercise 2: Challenge Problem - Designing a New GAN Application

### Foundation

Real product design requires selecting the right model family, collecting suitable data, and planning for failure cases before implementation.

This exercise applies GAN concepts to a game-development scenario.

### Build

Scenario:

You are a game developer building a tool where players draw a rough fantasy-character sketch (dragon, goblin, etc.), and AI converts it into polished concept art in the game's style.

Address the following design questions:

1. **Choose the Right GAN**

   From cGAN, CycleGAN, and StyleGAN, choose the best starting architecture for this sketch-to-art system.

   Justify your choice using the input-output relationship in this task.

2. **Dataset Requirements**

   What training data is needed?

   Would paired sketch/art examples be required, or could unpaired collections work? Explain your reasoning.

3. **Potential Challenges**

   Identify one or two major challenges in making the system practical and robust.

   Consider issues such as sketch quality variability, ambiguity in player intent, style consistency, and preserving key character attributes.

### Result

Your submission should present a coherent model choice, data strategy, and risk analysis for building a real sketch-to-concept-art pipeline.

## Exercise 2: Implementing Scaled Dot-Product Attention (90 minutes)

### Foundation

Scaled dot-product attention computes similarity between queries and keys, normalizes similarities into probabilities, and uses those probabilities to mix value vectors.

This exercise focuses on implementing each mathematical step directly in PyTorch.

### Build

Implement attention from scratch in PyTorch.

1. **Setup Tensors**

   Import torch and torch.nn.functional as F.

   Define random tensors for a batch of one sentence with sequence length 4 and embedding size 8:

   ```python
   import torch
   import torch.nn.functional as F

   # (batch_size, seq_len, d_k)
   queries = torch.randn(1, 4, 8)
   keys = torch.randn(1, 4, 8)
   values = torch.randn(1, 4, 8)
   ```

2. **Step 1 - Dot Product Scores**

   Compute dot products between queries and keys using a transpose on the last two dimensions of keys:

   ```python
   scores = queries @ keys.transpose(-2, -1)
   ```

   Print shape and verify it is `(1, 4, 4)`.

3. **Step 2 - Scaling**

   Get `d_k` from keys and divide scores by `sqrt(d_k)`.

4. **Step 3 - Softmax**

   Apply softmax over the last dimension (`dim=-1`) to get attention probabilities.

   Store in `attention_weights` and print the tensor.

5. **Step 4 - Weighted Sum of Values**

   Multiply `attention_weights` by `values` to get final output.

   Print shape and verify it is `(1, 4, 8)`.

6. **Encapsulate in a Function**

   Wrap full logic into:

   ```python

       ...
   ```

## Exercise 1: Conceptual Questions (30 minutes)

### Foundation

Reliable LLM agents depend on tool access, memory strategy, and transparent reasoning loops.

This exercise focuses on practical system design choices for real agent behavior in production-like scenarios.

### Build

Answer the following questions in markdown cells or comments.

1. **Tool Selection**

   You are designing an LLM travel-planning agent.

   List at least three essential tools (APIs) the agent should have.

2. **RAG vs. Fine-tuning for Memory**

   You are building a customer-support agent and the product documentation changes weekly.

   Would you fine-tune weekly or use RAG with a vector database?

   Explain pros and cons of both approaches for this specific update cadence.

   In ReAct, the model generates a `Thought` before an `Action`.

   Why is this important for reliability and debuggability?

## Exercise 2: Designing a ReAct Prompt (45 minutes)

### Foundation

ReAct combines explicit reasoning and tool use in a loop: Thought -> Action -> Observation -> ... -> Final Answer.

Prompt quality directly affects whether the model follows this structure reliably.

### Build

Write one complete prompt that instructs an LLM to solve this target question:

"Who is the director of the movie that won the Oscar for Best Picture in the same year that the movie 'Inception' was released?"

Available tools:

- `search(query: string)`: searches the web and returns a text snippet
- `get_movie_details(title: string)`: returns movie details, including director

Your prompt must include all of the following:

1. A clear instruction that the model is a helpful assistant that can use tools.

2. A tool section describing the tools and exact Action format, for example:

   `Action: search(query="...")`

3. One few-shot ReAct example for a different question, such as:

   "Who was the US president when the first iPhone was released?"

4. The final target question at the end of the prompt.

After writing the prompt, run it in an LLM interface and observe whether the model follows the Thought/Action/Observation loop to reach the answer.

### Result

Your submission should include your full prompt, a sample run transcript, and a short note on whether the ReAct structure was followed correctly.

## Exercise 3: Challenge Problem - Designing a RAG System

### Foundation

A RAG system combines retrieval over an external knowledge store with generation by an LLM.

Good RAG performance depends on ingestion quality, retrieval relevance, and prompt construction.

### Build

Design a conceptual RAG pipeline for a university "Study Buddy" agent with these sources:
- 50 PDF lecture notes
- 10 textbook chapters
- personal notes in text files

1. **Ingestion Process**

   Describe how you would prepare and store these materials in a vector database.

   Include practical steps such as text extraction, cleaning, chunking strategy, embeddings, metadata, and indexing.

2. **Retrieval Process**

   Given the query:

   "Can you explain the Bellman equation in the context of the cliff walking example from Lecture 4?"

   Explain how the system retrieves relevant chunks before LLM generation.

   Mention query embedding, similarity search, optional reranking, and context selection.

3. **Generation Process**

   Describe what final LLM prompt should contain.

   It should include retrieved context plus user question, with clear answer instructions and citation behavior.

### Result

Your answer should present a coherent end-to-end RAG design from ingestion to retrieval to generation for this course assistant use case.

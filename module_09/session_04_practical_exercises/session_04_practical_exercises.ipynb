## Exercise 2: Building a Simple GAN for MNIST (90 minutes)

### Foundation

A basic GAN for MNIST can be implemented with fully connected networks: a Discriminator that predicts real/fake probability and a Generator that maps latent noise to image space.

Training alternates updates: first improve Discriminator on real and fake samples, then improve Generator to fool Discriminator.

### Build

Implement and train a simple GAN on MNIST.

1. **Define the Networks**

   **Discriminator** (Sequential):
   - Input: flattened `28x28` image (`784`)
   - Hidden: 2-3 fully connected layers with `LeakyReLU`
   - Output: 1 neuron with `Sigmoid`

   **Generator** (Sequential):
   - Input: latent vector dimension (for example `100`)
   - Hidden: 2-3 fully connected layers with `LeakyReLU`
   - Output: `784` with `Tanh` activation

   Normalize MNIST images to `[-1, 1]` to match `Tanh` output range.

2. **Set Up Training**

   Define hyperparameters:
   - learning rate
   - batch size
   - number of epochs
   - latent dimension

   Create model instances, then define:
   - loss: `BCELoss`
   - optimizer for Discriminator (for example Adam)
   - optimizer for Generator (for example Adam)

3. **Write the Training Loop**

   For each epoch and each batch:

   **Train Discriminator**
   - Compute loss on real images with label `1`
   - Generate fake images from random `z` and compute loss with label `0`
   - Sum losses, backpropagate, update Discriminator

   **Train Generator**
   - Generate a fresh batch of fake images
   - Pass fake images through Discriminator
   - Compute Generator loss against label `1` (wants fakes judged real)
   - Backpropagate, update Generator

4. **Visualize Results**

   Periodically (for example each epoch), generate sample images, reshape to `28x28`, and display/save in a grid.

   Use these snapshots to observe training progression from noise-like outputs to digit-like outputs.

### Result

Your notebook should include working GAN code, training logs/loss trends, and periodic generated-image grids showing qualitative progress over epochs.

## Exercise 3: Challenge Problem - Thinking About Mode Collapse

### Foundation

Mode collapse is a common GAN failure where Generator outputs lose diversity and repeat only a few sample types.

Regularization techniques often focus on preventing the Discriminator from becoming overconfident too quickly.

### Build

1. Train your GAN for many epochs and inspect generated samples.

2. Check whether the Generator appears to collapse to only a few digit styles/classes.

3. Conceptual question:

   Label smoothing (or adding slight noise to Discriminator inputs/labels) is often used to fight collapse.

   Explain why making the Discriminator slightly less certain can provide better learning signals to the Generator and encourage broader mode exploration.

### Result

Your submission should include observations about collapse (or lack of it) and a clear explanation of why softening Discriminator confidence can stabilize training and improve diversity.

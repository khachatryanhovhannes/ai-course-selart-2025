## Exercise 2: Fine-Tuning a Pre-trained Model (Coding)

### Foundation
We use VGG16 pre-trained on ImageNet as a fixed feature extractor, then attach a custom binary classifier head for cats vs dogs.

This notebook focuses on model construction and analysis, assuming local image data is already organized.


```python
# 1) Setup: imports
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.applications.vgg16 import VGG16
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Flatten
```

```python
# 2) Load pre-trained base model (without top classifier)
base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(150, 150, 3)
)
```

```python
# 3) Freeze base model layers
for layer in base_model.layers:
    layer.trainable = False
```

```python
# 4) Add custom classification head
x = base_model.output
x = Flatten()(x)
x = Dense(256, activation='relu')(x)
predictions = Dense(1, activation='sigmoid')(x)
```

```python
# 5) Create and compile the final model
model = Model(inputs=base_model.input, outputs=predictions)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# 6) Analyze model structure
model.summary()
```

### Observation (Markdown Answer)

### Foundation
After running `model.summary()`, the output should show that only the new top layers are trainable.

### Build
- **Total params:** `16,812,353`
- **Trainable params:** `2,097,665`
- **Non-trainable params:** `14,714,688`

### Result
The trainable parameter count is much smaller because all VGG16 convolutional layers are frozen (`layer.trainable = False`), so training updates only the custom head (`Flatten -> Dense(256) -> Dense(1)`).

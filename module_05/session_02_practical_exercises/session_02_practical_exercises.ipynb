## Exercise 1: Activation Functions (Conceptual)

**Given inputs:** `[-5, -1, 0, 1, 5]`

### Foundation
- Sigmoid: `S(z) = 1 / (1 + e^(-z))`
- Tanh: `tanh(z)`
- ReLU: `max(0, z)`
- Leaky ReLU (a = 0.1): `z if z > 0 else 0.1*z`
- Swish: `Swish(z) = z * Sigmoid(z)`

### Build
- Sigmoid outputs (approx): `[0.0067, 0.2689, 0.5000, 0.7311, 0.9933]`
- Tanh outputs (approx): `[-0.9999, -0.7616, 0.0000, 0.7616, 0.9999]`
- ReLU outputs: `[0, 0, 0, 1, 5]`
- Leaky ReLU outputs: `[-0.5, -0.1, 0, 1, 5]`
- Swish outputs (approx): `[-0.0335, -0.2689, 0.0000, 0.7311, 4.9665]`

### Result
- **Zero-centered output:** `tanh` (symmetric around 0).
- **Computationally simplest:** `ReLU`.
- **Preferred for deep neural networks:** `ReLU` family (often `ReLU` or `Leaky ReLU` in hidden layers for stable and efficient training; `Swish` can also work well but is more expensive).


## Exercise 2: The Forward Pass (Tracing the Math)

**Given:**
- `X = [0.5, 0.3]`
- `W1 = [[0.2, 0.4], [0.7, 0.9]]`
- `B1 = [0.1, 0.2]`
- `W2 = [[0.6], [0.8]]`
- `B2 = [0.3]`
- Hidden activation: `ReLU`, output activation: `Sigmoid`

### Foundation
- `Z1 = X.W1 + B1`
- `A1 = ReLU(Z1)`
- `Z2 = A1.W2 + B2`
- `y_hat = Sigmoid(Z2)`

### Build
1. Hidden weighted sums:
   - `z1_1 = 0.5*0.2 + 0.3*0.7 + 0.1 = 0.1 + 0.21 + 0.1 = 0.41`
   - `z1_2 = 0.5*0.4 + 0.3*0.9 + 0.2 = 0.2 + 0.27 + 0.2 = 0.67`
   - `Z1 = [0.41, 0.67]`

2. Hidden activations (ReLU):
   - `A1 = [ReLU(0.41), ReLU(0.67)] = [0.41, 0.67]`

3. Output weighted sum:
   - `Z2 = 0.41*0.6 + 0.67*0.8 + 0.3`
   - `Z2 = 0.246 + 0.536 + 0.3 = 1.082`

4. Final prediction (Sigmoid):
   - `y_hat = 1 / (1 + e^(-1.082))`
   - `y_hat ~= 0.7469`

### Result
The final network output is **`y_hat ~= 0.747`**.


## Exercise 3: Architecture Design (Critical Thinking)

### Foundation
Task: multi-class classification for grayscale images (`Cat`, `Dog`, `Bird`), each image is `32 x 32`.

### Build
1. **Input Layer neurons:** `32 * 32 = 1024`
   - Because each pixel in a grayscale image contributes one input feature.

2. **Hidden layer activation:** `ReLU` (or `Leaky ReLU`)
   - Efficient computation, good gradient behavior in deep networks, and widely used in practice.

3. **Output Layer neurons:** `3`
   - One neuron per class: Cat, Dog, Bird.

4. **Output activation:** `Softmax`
   - Produces class probabilities that sum to 1 for multi-class single-label classification.

### Result
A suitable design is: `Input(1024) -> Hidden layers with ReLU/Leaky ReLU -> Output(3) with Softmax`.

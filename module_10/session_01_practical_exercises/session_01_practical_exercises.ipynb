**Total Time Estimate:** 90-120 minutes

**Objective:** To practice applying the foundational principles of AI ethics to real-world and hypothetical scenarios through short-form analysis and critical reasoning.


## Exercise 1: Analyzing a Real-World AI Controversy (45 minutes)

### Research
Recent article (within the last 1-2 years):
- AI-generated robocall impersonating U.S. President Joe Biden in New Hampshire (AP News, January 22, 2024): https://apnews.com/article/f3469ceb6dd613079092287994663db5

### Summarize
The AI system in this case was a voice-cloning generative AI model used to produce a synthetic audio message that sounded like President Biden. The call was distributed to New Hampshire voters before the presidential primary and told people to avoid voting in the primary, framing it as a strategic choice. Instead of supporting legitimate communication, the system was allegedly used for deceptive political messaging and possible voter suppression. The negative outcome was not just misinformation at scale, but direct erosion of trust in democratic communication channels, because recipients could not reliably distinguish authentic speech from fabricated speech.

### Analyze
This controversy conflicts with all five principles. Beneficence was violated because the deployment created social harm rather than public benefit. Non-maleficence was violated because the design and use of the model introduced foreseeable harm, including manipulation and civic confusion. Autonomy was violated because voters were nudged by deceptive identity claims, undermining informed, independent decision-making. Justice was violated because targeted misinformation can unequally affect groups with lower media literacy or fewer verification resources, creating unfair democratic participation conditions. Explicability was violated because recipients were not clearly informed that the audio was synthetic, and the origin, accountability chain, and decision process behind message generation and distribution were hidden.


## Exercise 2: The Ethical Matrix (45 minutes)

### Beneficence
**Benefit:** The assistant can improve student outcomes by creating a balanced study plan that adapts to workload and stress signals.

**Risk:** If optimization prioritizes grades only, the system may ignore broader well-being and reduce overall quality of life.

### Non-maleficence
**Benefit:** Early detection of burnout patterns can reduce harm by scheduling breaks and sleep before severe exhaustion occurs.

**Risk:** Continuous biometric monitoring may cause anxiety, self-surveillance pressure, or harmful over-reliance on algorithmic judgments.

### Autonomy
**Benefit:** Students can use recommendations as decision support, helping them make more informed choices about time management.

**Risk:** Automatic scheduling can become coercive if students feel unable to override the plan without penalties or guilt.

### Justice
**Benefit:** If widely accessible, the tool can provide high-quality study coaching to students who cannot afford private tutors.

**Risk:** Students without smart devices, stable internet, or compatible wearables may be excluded and left at a disadvantage.

### Explicability
**Benefit:** Transparent explanations for each schedule choice can help students understand why tasks, breaks, and sleep windows are recommended.

**Risk:** If the model is opaque, students cannot audit whether decisions come from valid evidence, biased assumptions, or flawed data.


## Exercise 3: Challenge Problem - Compliance vs. Ethics

The startup's product is legal in this scenario because it directly satisfies the requirements of the Productivity Mandate. If the law obligates large companies to deploy AI monitoring for productivity and the product implements those specified controls, the company is compliant in a formal legal sense.

Legal compliance, however, does not automatically make the product ethical. From an autonomy perspective, constant tracking of keystrokes, websites, and application time can undermine workers' control over their own professional behavior, especially when monitoring is mandatory and difficult to refuse. Even if employees are informed, forced consent under employment pressure is ethically weak. From a non-maleficence perspective, high-resolution surveillance can create harm through stress, fear of misinterpretation, reduced psychological safety, and possible misuse of personal behavioral data. A system may also misclassify legitimate work patterns, punishing deep thinking, caregiving interruptions, or accessibility needs. In that case, harm is not accidental; it is structurally built into how the product measures people.

As an engineer, personal responsibility includes more than implementing required features. I am responsible for identifying foreseeable harms, raising concerns, documenting risks, proposing safeguards, and refusing to normalize clearly abusive design choices when possible. Examples include data minimization, strict retention limits, human review for adverse decisions, opt-out paths where feasible, and transparent employee notification. The argument "I was just following the law" is not a sufficient ethical defense because laws can lag behind technology, and legality is a minimum threshold, not a complete standard for responsible engineering. Ethical practice requires independent judgment about human impact, not only procedural compliance.


A practical ethical approach is to treat compliance requirements as constraints, then still optimize for dignity, proportionality, and contestability. Proportionality asks whether the depth of monitoring is truly necessary for the stated productivity goal. Contestability asks whether employees can challenge inaccurate inferences and receive timely correction. Without these mechanisms, a compliant product can still become a tool of routine harm.

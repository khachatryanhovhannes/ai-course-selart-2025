## Exercise 2: Calculating IoU (30 minutes)

### Foundation

Intersection over Union (IoU) measures how much a predicted bounding box overlaps with the ground truth box. It is one of the most important calculations used to evaluate object detection quality.

### Build

This is a pen-and-paper or text editor calculation to solidify your understanding of IoU.

Consider an image that is `100 x 100` pixels.

- The Ground Truth bounding box for an object is `(x_min=20, y_min=20, x_max=80, y_max=70)`.
- Your model's Predicted bounding box is `(x_min=30, y_min=25, x_max=90, y_max=60)`.

Complete the following steps and show your work:

1. Calculate the area of the Ground Truth box.
2. Calculate the area of the Predicted box.
3. Calculate the coordinates of the intersection area, meaning the overlapping rectangle.
4. Calculate the area of the intersection.
5. Calculate the area of the union.

   ```text
   Area of Union = Area(A) + Area(B) - Area(Intersection)
   ```

6. Calculate the final IoU score.

   ```text
   IoU = Area of Intersection / Area of Union
   ```

### Result

Your final answer should include the ground truth area, predicted area, intersection coordinates, intersection area, union area, and IoU score.

## Exercise 1 — Trace Alpha–Beta Pruning (Conceptual)

We consider the same game tree as in the previous session.  
The root **A** is a **MAX** node.

### 1. Game tree

```text
A (MAX)
├── B (MIN)
│   ├── E = 3
│   └── F = 5
├── C (MIN)
│   ├── G = 6
│   ├── H = 9
│   └── I = 7
└── D (MIN)
    ├── J = 0
    └── K = 1
```

We apply **Minimax with Alpha–Beta pruning**, exploring children **left-to-right**.

Initial values at the root:  
- α (alpha) = −∞  
- β (beta) = +∞  

---

### 2. Depth-first trace with α and β

#### Step 1 — Node A (MAX)

- At **A (MAX)**:  
  - α = −∞, β = +∞  
  - Children (left to right): **B, C, D**

---

#### Step 2 — Subtree of B (MIN)

1. Enter **B (MIN)** with:
   - α = −∞, β = +∞

2. Explore **E**:
   - **E** is a terminal leaf with value **3**  
   - At **B**:  
     - current value = min(+∞, 3) = **3**

     - No pruning yet because β = 3 > α = −∞

3. Explore **F**:
   - **F** is a terminal leaf with value **5**  
   - At **B**:  
     - current value = min(3, 5) = **3**  
     - β = min(3, 5) = **3**  

4. Result for **B**:
   - **B(MIN) = 3**

5. Back to **A (MAX)**:
   - current value at A = max(−∞, 3) = **3**  
   - α = max(−∞, 3) = **3**  
   - β at A remains +∞  
   - No pruning at A yet.

#### Step 3 — Subtree of C (MIN)

1. Enter **C (MIN)** with:
   - α = 3, β = +∞

2. Explore **G**:
   - **G** is a terminal leaf with value **6**

   - **H** is a terminal leaf with value **9**

4. Explore **I**:
   - **I** is a terminal leaf with value **7**

## Exercise 2 — The Power of Move Ordering (Critical Thinking)

We now consider the effect of **move ordering** at the root **A (MAX)**.

The true Minimax values of the three subtrees are:
- B(MIN) = 3  
- C(MIN) = 6  
- D(MIN) = 0  

---

### 1. What if we explore A’s children in the order D, then C, then B?

With the order **D → C → B**:

- First we explore **D** and get value **0**, so at A we have α = 0.  
- Then we explore **C** and get value **6**, so α becomes **6**.  
- Finally we explore **B** with α = 6:
  - At B (MIN), after evaluating leaf **E = 3**, β becomes 3.
  - Since β = 3 ≤ α = 6, the second leaf **F** is **pruned**.

So in this new order:

- We still prune **exactly one leaf**, but now it is **F** instead of **K**.  
- **Amount of pruning:** *the same* as in the left-to-right order (one terminal node skipped).

---

### 2. More, less, or the same pruning? (Short reasoning)

- The order **D, C, B** gives **the same amount of pruning** as the original left-to-right order:
  - In both cases we avoid evaluating **one** terminal node.  
- The difference is **where** the pruning happens:
  - Original order B, C, D → pruned **K** under D  
  - New order D, C, B → pruned **F** under B

---

### 3. Ideal ordering of moves for A (MAX) for maximum pruning

For Alpha–Beta pruning, the **ideal ordering for MAX** is to explore children **from the best (highest value) to the worst**.  

Here, the subtree Minimax values are:

- C = 6  (best)  
- B = 3  
- D = 0  (worst)

So an ideal ordering for **A (MAX)** is:

- **C first, then B, then D** (i.e., **C → B → D**)  
  - or **C → D → B** (also starting with the best child C)

In such an ordering:

- A quickly reaches a **high α value** after exploring C.  
- This higher α allows **more pruning** in the remaining subtrees (B and D).  
- In this specific tree, starting with C allows us to prune **two** leaves instead of just one.

**Conclusion:**  
> For maximum pruning, the MAX player at node **A** should explore children in **descending order of their Minimax values**, i.e. **C first, then B, then D**.

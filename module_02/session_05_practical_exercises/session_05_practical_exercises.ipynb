## Exercise 1: Trace the Minimax Algorithm

We will compute the Minimax value of each internal node step by step, starting from the leaves and moving up to the root.

---

### 1. Leaf Utilities (Given)

We start from the terminal nodes, whose values are already known:

- `E = 3`
- `F = 5`
- `G = 6`
- `H = 9`
- `I = 7`
- `J = 0`
- `K = 1`

These are fixed utilities from the MAX player's perspective.  
We now use them to compute the values of the internal nodes.

---

### 2. Values of MIN Nodes at Level 1 (B, C, D)

All three nodes **B, C, D** are MIN nodes, so each of them will choose the **smallest** value among its children.

#### 2.1 Node B (MIN)

Children of B: **E** and **F**

Since **B** is a **MIN** node:

```
Value(B) = min(3, 5) = 3
```

So:

**Value(B) = 3**

---

#### 2.2 Node C (MIN)

Children of C: **G**, **H**, and **I**

Since **C** is a **MIN** node:

```
Value(C) = min(6, 9, 7) = 6
```

So:

**Value(C) = 6**

---

#### 2.3 Node D (MIN)

Children of D: **J** and **K**

Since **D** is a **MIN** node:

```
Value(D) = min(0, 1) = 0
```

So:

**Value(D) = 0**

At this point, we have the values for all MIN nodes at Level 1:

- `Value(B) = 3`
- `Value(C) = 6`
- `Value(D) = 0`

---

### 3. Value of the Root Node A (MAX)

Node **A** is a MAX node. Its children are **B, C, and D**, with the values we just computed:

Since **A** is a **MAX** node:

```
Value(A) = max(3, 6, 0) = 6
```

So the final Minimax value of the root node is:

**Value(A) = 6**

---

### 4. Optimal First Move for the MAX Player

Now we compare the outcomes from each possible first move:

- If MAX goes from **A to B**, the result is **3**.
- If MAX goes from **A to C**, the result is **6**.
- If MAX goes from **A to D**, the result is **0**.

Because the MAX player wants to **maximize** the outcome, the best choice is the move that leads to value **6**.

Therefore, the optimal first move is:

**Move from A to C.**

---

### 5. Final Answers

1. **Final value of the root node (A):**  
   → `Value(A) = 6`

2. **Optimal first move for the MAX player:**  
   → The MAX player should move from **A** to **C**.

## Exercise 2: Designing a Utility Function (Creative Thinking)

In this exercise, we design an **evaluation (utility) function** for a non-terminal game state of **Tic-Tac-Toe**, from the perspective of the **MAX player (X)**.

---

## 1. Problem Recap

The Minimax algorithm can only work with terminal states — meaning when the game is completely finished.  
However, in larger games (like chess or tic-tac-toe with depth limit), we cannot explore the entire game tree because it becomes too large.

To solve this, we use an **evaluation function** (also called a heuristic function) that estimates how good a non-terminal state is for the MAX player.

---

## 2. Terminal State Utilities

We define simple utility values for terminal states:

- **Win for X (MAX player): +1**
- **Win for O (MIN player): -1**
- **Draw:** 0

But since we are dealing with *non-terminal* states too, we need a way to **approximate** these outcomes before the game ends.

---

## 3. Designing the Evaluation Function

A good evaluation function should measure how close each player is to winning.

We can think in terms of “lines” (rows, columns, and diagonals).  
Each player tries to make 3 in a row — so we can measure the potential of each line.

---

### Example Evaluation Function

```
Evaluation(state) =
   (Number of lines where X can still win)
 - (Number of lines where O can still win)
```

This is a simple version, but we can make it more detailed.

---

### Improved Version (Feature-Based)

We can count *patterns* that show partial progress toward winning.

Let’s define weights based on the number of X’s and O’s in a line:

| Line Pattern (in 3 cells) | Meaning | Contribution to Score |
|----------------------------|----------|------------------------|
| `X X X` | X has already won | +100 |
| `O O O` | O has already won | -100 |
| `X X _` | X is close to winning | +10 |
| `O O _` | O is close to winning | -10 |
| `X _ _` | X has small advantage | +1 |
| `O _ _` | O has small advantage | -1 |

Then the evaluation function becomes:

```
Evaluation(state) =
   100 * (number of X 3-in-a-row)
 - 100 * (number of O 3-in-a-row)
 + 10  * (number of X 2-in-a-row with an empty cell)
 - 10  * (number of O 2-in-a-row with an empty cell)
 + 1   * (number of X single-in-a-row)
 - 1   * (number of O single-in-a-row)
```

This version rewards X for being closer to winning and penalizes for O’s progress.

---

## 4. Intuitive Explanation

- If X is one move away from winning, the score becomes strongly **positive**.
- If O is one move away from winning, the score becomes strongly **negative**.
- If both players are equally close, the score stays near **0**.

This makes the evaluation function behave like an intelligent “guess” — it predicts who is more likely to win if both play optimally from this state.

---

## 5. Example

Example board state:

```
X | O | X
O | X | _
_ | O | _
```

Now we can count:

- X has one “two-in-a-row” with a chance to win → +10
- O also has one “two-in-a-row” → -10
- X has some additional single marks → +2
- O has one additional single → -1

So:

```
Evaluation(state) = +10 - 10 + 2 - 1 = +1
```

This means: **Slight advantage for X**.

---

## 6. Final Answer (Summary)

- **Evaluation function goal:** Estimate how favorable a non-terminal Tic-Tac-Toe board is for X (MAX).
- **Core idea:** Count the number of potential winning opportunities for X and O.
- **Better version:** Use feature weights for 3-in-a-row, 2-in-a-row, and single marks.
- **Example output:** Positive → advantage for X, Negative → advantage for O, Zero → balanced position.
